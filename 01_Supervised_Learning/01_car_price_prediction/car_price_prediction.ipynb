```python
X=df.drop(columns=['price'],axis=1)
y=df['price']
```

```python
X
```

```python
col=['model','transmission','fuelType']
x_one_encoded=pd.get_dummies(X,col,drop_first=True)
x_one_encoded=x_one_encoded.astype(int)
x_one_encoded
```

```python
from sklearn.preprocessing import LabelEncoder
encode=LabelEncoder()
xlabel_encoded=X.copy()
for c in col:
  xlabel_encoded[c]=encode.fit_transform(xlabel_encoded[c])
```

```python
xlabel_encoded
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score
```

```python
x_train,x_test,y_train,y_test=train_test_split(x_one_encoded,y,test_size=0.2,random_state=42)
model=LinearRegression()
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
y_pred
```

```python
r2=r2_score(y_test,y_pred)
r2
```

```python
n=x_test.shape[0]
p=x_test.shape[1]
adj_r2=1-((1-r2)*(n-1)/(n-p-1))
adj_r2
```

```python
x_train,x_test,y_train,y_test=train_test_split(xlabel_encoded,y,test_size=0.2,random_state=42)
model2=LinearRegression()
model2.fit(x_train,y_train)
y_pred=model2.predict(x_test)
y_pred
```

```python
y_test
```

```python
r22=r2_score(y_test,y_pred)
r22
```
